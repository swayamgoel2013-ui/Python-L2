**API** = `Application Programming Interface`

An API allows different software applications to talk to each other. It acts as a middleman, taking a request from one program, delivering it to another, and bringing the response back.

- Lets look at some examples to understand the use of APIs


**Weather API**

```python
import requests
import json

latitude, longitude = 28.5424263, 77.1829994
response = requests.get(f"https://api.open-meteo.com/v1/forecast?latitude={latitude}&longitude={longitude}&current_weather=True")
print(response)

json_data = json.loads(response.text)

with open("output.json", "w", encoding="utf-8") as f:
    json.dump(json_data, f, ensure_ascii=False, indent=4)
```

**Pokemon API**

```python
import requests
import json

pokemon_name = "pikachu"
response = requests.get(f"https://pokeapi.co/api/v2/pokemon/{pokemon_name}")
json_data = json.loads(response.text)

with open("output.json", "w", encoding="utf-8") as f:
    json.dump(json_data, f, ensure_ascii=False, indent=4)
```

```python
ability_name = json_data["abilities"][0]["ability"]["name"]
print(f"{ability_name = }")

ability_url = json_data["abilities"][0]["ability"]["url"]
print(f"{ability_url = }")

response = requests.get(ability_url)
ability_data = json.loads(response.text)

with open("output.json", "w", encoding="utf-8") as f:
    json.dump(ability_data, f, ensure_ascii=False, indent=4)
```

**Currency Exchange Rate API**

```python
import requests
import json


convert_from = "USD"
response = requests.get(f"https://api.exchangerate-api.com/v4/latest/{convert_from}")
json_data = json.loads(response.text)

with open("output.json", "w", encoding="utf-8") as f:
    json.dump(json_data, f, ensure_ascii=False, indent=4)

json_data["rates"]["INR"]
```

```python
import requests
import json

amount       = float(input("Enter the amount: "))
convert_from = input("Enter source currency: ")
convert_to   = input("Enter target currency: ")

url = f"https://api.exchangerate-api.com/v4/latest/{convert_from}"
response = requests.get(url)
json_data = json.loads(response.text)
    
multiplier = json_data["rates"][convert_to]
answer = amount * multiplier

print(f"{amount} {convert_from} is equal to {answer} {convert_to}")
```

**Dictionary API**

```python
import requests
import json

word = "computer"
response = requests.get(f"https://api.dictionaryapi.dev/api/v2/entries/en/{word}")

json_data = json.loads(response.text)
with open("output.json", "w", encoding="utf-8") as f:
    json.dump(json_data, f, ensure_ascii=False, indent=4)
```

```python
import requests
import json

word = "computer"
response = requests.get(f"https://api.dictionaryapi.dev/api/v2/entries/en/{word}")

json_data = json.loads(response.text)

list_of_definitions = json_data[0]["meanings"][0]["definitions"]

count = 0
for definition in list_of_definitions:
    count += 1
    print(f">> [{count}] {definition['definition']}")
```

**Github API**

```python
import requests
import json

username = "swayamgoel2013-ui"
url = f"https://api.github.com/users/{username}"

response = requests.get(url)
json_data = response.json()

with open("output.json", "w", encoding="utf-8") as f:
    json.dump(json_data, f, ensure_ascii=False, indent=4)

if response.status_code == 200:
    data = response.json()
    
    print(f"Username: {data["login"]} / {data["name"]}")
    print(f"Public Repos: {data["public_repos"]}")
    print(f"Followers: {data["followers"]}")
    print(f"Following: {data["following"]}")
else:
    print("Error:", response.status_code)
```
