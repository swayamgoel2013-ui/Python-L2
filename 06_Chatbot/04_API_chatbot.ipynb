
- Then, in your Python code, import the `load_dotenv` function from the `dotenv` module and call the `load_dotenv` function, as follows:

````python
import os
from dotenv import load_dotenv

load_dotenv(override=True) 
API_KEY = os.getenv("API_KEY")

# If you can see your API key in the output, you are good to go!
````

## Final Code

````python
# install the required libraries
import os
from dotenv import load_dotenv
import google.generativeai as genai

# load the API key from the .env file
load_dotenv(override=True) 
API_KEY = os.getenv("API_KEY")

# configure genai
genai.configure(api_key=API_KEY)

# set up the model, "gemini-2.5-flash" is a good model to start with
model = genai.GenerativeModel(model_name="gemini-2.5-flash")

# start a conversation
chat = model.start_chat(history=[])

# intro message
print("Welcome to the Gemini Chatbot!")

# conversation loop
while (True):
    # user input
    request = input("Enter your prompt: ")
    # breaking condition
    if(request == "bye"):
        break
    # update chat
    response = chat.send_message(request).text
    print(f"\nYou >> {request}")
    print(f"\nBot >> {response}")

# closing message
print("\nGoodbye! Have a nice day!")
````
