```python
import pandas as pd
```

```python
df = pd.read_csv("dataset.csv")
df
```

```python
type(df.head(3))
```

```python
flying_mons = df.loc[ df['HP']>100, ['species_ID', 'HP'] ].reset_index(drop=True)
flying_mons
```

```python
physical_atkrs = df.loc[ df['phAtk']>df['spAtk'], : ].reset_index(drop=True)
physical_atkrs
```

```python
offensive = df.loc[ (df['phAtk'] + df['spAtk'])>(df['phDef'] + df['spDef']), : ].reset_index(drop=True)
offensive
```

```python
flying_mons = df.loc[ (df['HP']>80) | (df['speed']>80) ].reset_index(drop=True)
flying_mons
```

```python
# name of the fastest pokemon
fast = df.sort_values(by=['speed'], ascending=[False]).reset_index(drop=True)
fastest = fast.loc[0, 'species_ID']
fastest
```

```python
# count of normal type pokemons
normies = df.loc[ (df['type_1'] == "NORMAL") | (df['type_2'] == "NORMAL") ].reset_index(drop=True)
len(normies)
```

```python
# name of the fastest fire type pokemon
fire_mons = df.loc[ (df['type_1'] == "FIRE") | (df['type_2'] == "FIRE") ].reset_index(drop=True)
fast_firemon = fire_mons.sort_values(by=['speed'], ascending=[False]).reset_index(drop=True)
fastest = fast_firemon.loc[0, 'species_ID']
fastest
```

```python
ground_type = df.loc[(df['type_1'] == "GROUND") | (df['type_2'] == "GROUND")].reset_index(drop=True)

hp_ground_type = ground_type.loc[(df['HP']>50)].reset_index(drop=True)

len(hp_ground_type)
```
