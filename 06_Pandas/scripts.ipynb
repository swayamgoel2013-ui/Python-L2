````python
# importing pandas in this notebook
import pandas as pd
````

### Creating a dataframe form (2D) nested structures

````python
# list of lists
data = [
    ['Alex', 10],
    ['Bob', 12],
    ['Clarke', 13]
]

df = pd.DataFrame(data)
df
````

````python
# list of dict
data = [
    {'Name': 'Alex',  'Age': 10},
    {'Name': 'Bob',   'Age': 12},
    {'Name': 'Clarke','Age': 13},
]

df = pd.DataFrame(data)
df
````

````python
# dict of lists
data = {
    'Name':  ['Alex', 'Bob', 'Clarke'],
    'Age':   [10, 12, 13],
}

df = pd.DataFrame(data)
df
````

````python
# dict of dict
data = {
    'a': {'Name': 'Alex',  'Age': 10},
    'b': {'Name': 'Bob',   'Age': 12},
    'c': {'Name': 'Clarke','Age': 13},
}

df = pd.DataFrame(data)
df
````

````python
# dict of dict
data = {
    'Name': {'a': 'Alex',  'b': 'Bob', 'c': 'Clarke'},
    'Age': {'a': 10, 'b': 12, 'c':13},
}

df = pd.DataFrame(data)
df
````

### To access a single specific row / column / cell

Syntax:

```Python
cell = df.loc[ row_index , col_name ]
row  = df.loc[ row_index , : ]
col  = df.loc[ : , col_name ]
```

````python
# to get a single specific row
row_index = 1
row = df.loc[ row_index , : ]
print(type(row))

row
````

````python
# to get a single specific column
col_name = 'HP'
col = df.loc[ : , col_name ]
print(type(col))

row
````

````python
# to get single specific cell
row_index = 1
col_name = 'HP'
cell = df.loc[ row_index , col_name ]
print(type(cell))

cell
````

````python
# to get single specific cell
row_index = 1
col_name = 'species_ID'
cell = df.loc[ row_index , col_name ]
print(type(cell))

cell
````

### To access multiple rows / columns / cells

Syntax:

```Python
cells = df.loc[ row_filter , col_filter ]
rows  = df.loc[ row_filter , : ]
cols  = df.loc[ : , col_filter ]
```

````python
# to get multiple specific rows

rows = df.loc[ [0, 1, 2] , : ]
print(type(rows))

rows
````

````python
# to get multiple specific cols

cols = df.loc[ : , ['species_ID', 'type_1', 'type_2'] ]
print(type(cols))

cols
````

````python
# to get multiple specific cells

cells = df.loc[ [0, 1, 2] , ['species_ID', 'type_1', 'type_2'] ]
print(type(cells))

cells
````

### To filter rows based on column values

Syntax:

```Python
row_condition = df[column_name] > value
filtered_rows = df.loc[ row_condition ].reset_index(drop=True)
```

````python
# to get rows based on some condition (Logical AND)

row_condition = df["HP"] > 100
filtered_rows = df.loc[ row_condition ].reset_index(drop=True)
print(type(filtered_rows))

filtered_rows
````

````python
# to get rows based on some condition (Logical OR)

row_condition = (df["speed"] < 20) | (120 < df["speed"])
filtered_rows = df.loc[ row_condition ].reset_index(drop=True)
print(type(filtered_rows))

filtered_rows
````

````python
# to get rows based on some condition

row_condition = (80 < df["speed"]) & (df["speed"] < 100)
filtered_rows = df.loc[ row_condition ].reset_index(drop=True)
print(type(filtered_rows))

filtered_rows
````

filtered_rows = df.sort_values(by=[column_names], ascending=[bool_values]).reset_index(drop=True)

### Practice problems

Q: Name of the fastest pokemon

````python
speed_order = df.sort_values(by=['speed'], ascending=[False]).reset_index(drop=True)
fastest = speed_order.loc[0, 'species_ID']
fastest
````

Q: Count of normal type pokemons

````python
normal_mons = df.loc[ (df['type_1'] == "NORMAL") | (df['type_2'] == "NORMAL") ].reset_index(drop=True)
len(normal_mons)
````

Q: Name of the fastest fire type pokemon

````python
fire_mons = df.loc[ (df['type_1'] == "FIRE") | (df['type_2'] == "FIRE") ].reset_index(drop=True)
fast_firemon = fire_mons.sort_values(by=['speed'], ascending=[False]).reset_index(drop=True)
fastest_firemons = fast_firemon.loc[0, 'species_ID']
fastest_firemons
````

Q: Count of ground type mons having HP > 50

````python
ground_mons = df.loc[(df['type_1'] == "GROUND") | (df['type_2'] == "GROUND")].reset_index(drop=True)
healthy_ground_mons = ground_mons.loc[(df['HP']>50)].reset_index(drop=True)
len(healthy_ground_mons)
````
