```python
import pandas as pd
```

```python
# list of lists
data = [
    ['Alex', 10],
    ['Bob', 12],
    ['Clarke', 13]
]

df = pd.DataFrame(data)
df
```

```python
# list of dict
data = [
    {'Name': 'Alex',  'Age': 10},
    {'Name': 'Bob',   'Age': 12},
    {'Name': 'Clarke','Age': 13},
]

df = pd.DataFrame(data)
df
```

```python
# dict of lists
data = {
    'Name':  ['Alex', 'Bob', 'Clarke'],
    'Age':   [10, 12, 13],
}

df = pd.DataFrame(data)
df
```

```python
# dict of dict
data = {
    'a': {'Name': 'Alex',  'Age': 10},
    'b': {'Name': 'Bob',   'Age': 12},
    'c': {'Name': 'Clarke','Age': 13},
}

df = pd.DataFrame(data)
df
```

```python
# dict of dict
data = {
    'Name': {'a': 'Alex',  'b': 'Bob', 'c': 'Clarke'},
    'Age': {'a': 10, 'b': 12, 'c':13},
}

df = pd.DataFrame(data)
df
```
