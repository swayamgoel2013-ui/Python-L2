---
**Object Oriented Programming (O.O.P.)**
- The core entities of OOP are **classes** and **objects**.
---

**Classes**

- Think of classes as user-defined data types.
- When you define a class, you essentially define how this new data type is going to be like
- Mainly you define two things:
    - what kind of data it is going to store.
    - what kind of operations it is going to support.
---

**Objects**

- Think of objects as variables of these user-defined data types.
- When you initialize an object, you have created a variable that is going to store data and support operations as defined in the corresponding class.
- The data that an object stores are called its _attributes_.
- The operations that an object supports are called its _methods_.
---

````python
# Lets create a class to store Time as an object

class Time:
    def __init__(self, hour, minute):   # constructor: special function that is called when an object is created
        self.hour = hour                # a Time object is going to store two int values: hour and minute
        self.minute = minute

# main starts here
time1 = Time(12, 30)                    # creating an object of Time class

print(f"{time1 = }")                    # trying to print the object of Time class
print(f"{time1.hour   = }")               # trying to print the value of hour attribute of Time class
print(f"{time1.minute = }")             # trying to print the value of minute attribute of Time class
````

**Interaction of objects with python features**

- One of the first interactions that you may think of is to print an object using `print()` function.
- If try to print an object using print function you won't see anything meaningful.
- To print it properly, you need to define the `__repr__` method.
- The `__repr__` method is a special dunder method (just like `__init__` method).
- `__repr__` method defines how your object should be represented/printed.
- You can decide hwo descriptive you want it to be.
What to hide, what to show, in what order, in what format, everything.
- These methods are meant to serve unique purposes. More about them later.

````python
class without_repr:               # class without `__repr__` method
    def __init__(self, val):
        self.val = val
    
class with_repr:                  # class with `__repr__` method
    def __init__(self, val):
        self.val = val
    def __repr__(self):
        return f"value = {self.val}"    # you can format it the way you want

# main starts here
object_1 = without_repr(10)
print(object_1)

object_2 = with_repr(10)
print(object_2)
````

````python
class Time:
    def __init__(self, hour, minute):
        self.hour = hour
        self.minute = minute
    def __repr__(self):
        return f"{self.hour:02}:{self.minute:02}"

# main starts here
time1 = Time(6, 7)      # creating an object of Time class
print(f"{time1 = }")    # printing the object of Time class
````

**Interaction between classes**

- One reason to use OOP is to mimic real-life objects using programming.
- After creating objects and defining their attributes, the next thing we need to do is to define how will these objects are going to interact with each other.
- The objects may interact with other objects of the same class as well as objects of other classes.
- To define these interactions we have to define some **methods**.

- Think of methods as functions that are defined inside a class.
- A method defined inside a class can only be used by objects of that class, or the class itself.

````python
# Class to define a period of time
class TimePeriod:
    def __init__(self, minutes):
        self.minutes = minutes
    def __repr__(self):
        return f"{self.minutes} minutes"

# Class to define a point in time
class Time:
    def __init__(self, hour, minute):
        self.hour = hour
        self.minute = minute
    def __repr__(self):
        return f"{self.hour:02}:{self.minute:02}"
    
    # this method calculates the TimePeriod between two times
    # interaction between two objects of the same class
    def diff(self, other):
        total_minutes_1 = (self.hour * 60) + self.minute
        total_minutes_2 = (other.hour * 60) + other.minute
        period_in_minutes = total_minutes_1 - total_minutes_2
        return TimePeriod(period_in_minutes)
    
    # this method calculates the time after given time period
    # interaction between two objects of different classes
    def after(self, period):
        total_minutes_1 = (self.hour) * 60 + self.minute
        period_in_minutes = period.minutes
        total_minutes_2 = total_minutes_1 + period_in_minutes
        hour   = total_minutes_2 // 60
        minute = total_minutes_2 % 60
        return Time(hour, minute)


# main starts here

# Example 1: The class started at 04:21 and the current time is 05:14.
# I want to calculate how much time has passed
start_time = Time(16, 21)
curr_time = Time(17, 14)
time_passed = curr_time.diff(start_time)
print(f"Time passed: {time_passed}")

# Example 2: The class started at 04:21 and its supposed duration is 100 minutes.
# I want to calculate the end time for the class
start_time = Time(16, 21)
duration = TimePeriod(100)
end_time = start_time.after(duration)
print(f"End time: {end_time}")
````
