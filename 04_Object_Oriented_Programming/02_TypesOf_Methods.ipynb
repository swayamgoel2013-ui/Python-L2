**Instance & Class members**
- Instance members are those members _(attributes and methods)_ which are associated with each instance _(object)_ of a class.
- Class members are those members _(attributes and methods)_ which are shared by each instance _(object)_ of a class.

```python
class Student:
    # CLASS ATTRIBUTES : common for all objects of this class
    count = 0

    def __init__(self, name, marks):
        # INSTANCE ATTRIBUTES : unique for all objects of this class
        self.name = name
        self.marks = marks
        self.incrCounter()

    # CLASS METHOD : common for all objects of this class
    #                can only access class members and not instance members
    #                has `cls` as the first argument
    @classmethod
    def incrCounter(cls):
        cls.count += 1
    
    @classmethod
    def showCounter(cls):
        print(f"Total: {cls.count}")

    # INSTANCE METHOD : unique for all objects of this class
    #                   can access both class members and instance members
    #                   has `self` as the first argument
    def show_details(self):
        percent = self.percentage(self.marks, total_marks=150)
        print(f"Name : {self.name}")
        print(f"Marks: {self.marks}")
        print(f"Perc : {percent:.2f}%")
        print()

    # STATIC METHOD : common for all objects of this class
    #                 can access neither class members nor instance members 
    #                 used for making utility function that doesn't require access
    @staticmethod
    def percentage(marks_obtained, total_marks):
        return (marks_obtained/total_marks)*100

# main starts here
s1 = Student('Swayam', 87)
s2 = Student('Siddhu', 97)
s3 = Student('Rohan', 92)

s1.show_details()
s2.show_details()
s3.show_details()

Student.showCounter()
s1.showCounter()
s2.showCounter()
s3.showCounter()
```

```python
class Family_Member:
    room_count = 4
    def __init__(self, name, age):
        self.name = name
        self.age = age

# room_count is a class attribute hence we can access it via the class
# even if we don't have created any objects yet
print("initially when there's no member")
print(f"{Family_Member.room_count = }")
print()

# once we create an object of the class we can access it in two ways:
# - directly via the class
# - via the object (this will still refer to the class attribute as the objects doesn't have their own copy)
swayam = Family_Member("Swayam", 13)
print("after adding the first member")
print(f"{Family_Member.room_count = }")
print(f"{swayam.room_count            = }")
print()

# if we create multiple objects, we can access the class attribute via the class or any of its objects
# all of these will refer to the same class attribute, as objects don't have their own copies
siddhu = Family_Member("Siddhu", 20)
print("after adding the second member")
print(f"{Family_Member.room_count = }")
print(f"{swayam.room_count            = }")
print(f"{siddhu.room_count            = }")
print()

# if we update this value via the class, it will update the class attribute
# and the change will reflect in all the objects, as objects don't have their own copies
Family_Member.room_count = 5
print("after updating the class attribute via the class")
print(f"{Family_Member.room_count = }")
print(f"{swayam.room_count            = }")
print(f"{siddhu.room_count            = }")
print()

# if we update this value via any of the objects, it will create a separate copy of this for that object
# teh change will not reflect in other objects, as the original class attribute is unchanged
siddhu.room_count = 6
print("after updating the object attribute via an object (for siddhu)")
print(f"{Family_Member.room_count = }")
print(f"{swayam.room_count            = }")
print(f"{siddhu.room_count            = }")
print()

# if we now update this value via the class, it will update the class attribute
# and the change will reflect only for those objects that don't have their own copy
# and for the objects having their own copies, they will remain unchanged
Family_Member.room_count = 7
print("after updating the class attribute via the class again")
print(f"{Family_Member.room_count = }")
print(f"{swayam.room_count            = }")
print(f"{siddhu.room_count            = }")
print()
```
